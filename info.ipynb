3. **Automated CI pipeline:**  PR -> Event -> Workflow is Triggered
    1. once the pull request is open okay a CI pipeline is triggered
    2. This pipeline builds the application and runs all test cases.
    3. If the pipeline passes, the PR is approved and merged in the main branch.

4. **Continuous Deployment**
    1. Upon merging the PR, a CD pipeline is triggered.
    2. The application is automatically deployed to the Dev or Staging Environment.
    3. deploy to the production environment
